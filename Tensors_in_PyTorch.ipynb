Mathematically this looks like:

$$
\begin{align}
y &= f(w_1 x_1 + w_2 x_2 + b) \\
y &= f\left(\sum_i w_i x_i +b \right)
\end{align}
$$

With vectors this is the dot/inner product of two vectors:

$$
h = \begin{bmatrix}
x_1 \, x_2 \cdots  x_n
\end{bmatrix}
\cdot
\begin{bmatrix}
           w_1 \\
           w_2 \\
           \vdots \\
           w_n
\end{bmatrix}
$$

```python
import torch
```

```python
def activation(x):
  """
  Sigmoid activation funtion

  Arguments
  x: torch.Tensor
  """
  return 1/(1+torch.exp(-x))
```

Single layer neural network


```python
# Generate some synthetic data

torch.manual_seed(7)
features = torch.randn((1, 5))
weights = torch.rand_like(features)
bias = torch.randn((1,1))
```

```python
print(f"Features: {features}")
print(f"Weights: {weights}")
print(f"Bias: {bias}")
```

```python
# Calculating the output of the network using torch.sum()

y = activation(torch.sum(features * weights) + bias)
print(f"y = {y}")
```

```python
# Calculating the output of the network using matrix multiplication

y = activation(torch.mm(features, weights.view(5, 1)) + bias)
print(f"y = {y}")
```

Multi layer neural network

```python
## Generate some data
torch.manual_seed(7)

# Features are 3 random normal variables
features = torch.randn((1, 3))

# Define the size of each layer in the network
n_input = features.shape[1]
n_hidden = 2
n_output = 1

# Weights for inputs to hidden layer
w1 = torch.randn(n_input, n_hidden)
# Weights for hidden to output layer
w2 = torch.randn(n_hidden, n_output)

# Bias for hidden and output layer
b1 = torch.randn((1, n_hidden))
b2 = torch.randn((1, n_output))
```

```python
## Calculating the network output using matrix multiplication

h = activation(torch.mm(features, w1) + b1)
output = activation(torch.mm(h, w2) + b2)

print(f"y = {output}")
```

NumPy to Torch and back


```python
import numpy as np
```

```python
# Genarate some data

a = np.random.rand(4, 3)
print(a)
```

```python
# Convert numpy array to tensors

b = torch.from_numpy(a)
print(b)
```

```python
# Convert tensors to numpy array

b.numpy()
```

The memory is shared between the Numpy array and Torch tensor, so if you change the values in-place of one object, the other will change as well.

```python
b.mul_(2)
```
